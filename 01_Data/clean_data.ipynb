```python
import pandas as pd
```

```python
df = pd.read_csv('retail_sales.csv', sep=';')

df.head()
```

```python
#missing values
missing_values = df.isnull().sum()
missing_values
```

```python
df['City'] = df['City'].fillna('Unknown')
```

```python
df['Payment_Method'] = df['Payment_Method'].fillna('Unkown')
```

```python
df = df[df['Payment_Method'] != 'Unkown']

# Verify it worked
print(df['Payment_Method'].value_counts())
```

```python
df = df.dropna(subset=['Customer_ID', 'Quantity', 'Unit_Price_ZAR', 'Discount_Percent', 'Unit_Cost_ZAR', 'Revenue_ZAR', 'Profit_ZAR'])
```

```python
missing_values = df.isnull().sum()
missing_values
```

```python
df.head()
```

```python
data_types = df.dtypes
data_types
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])



df['Quantity'] = pd.to_numeric(df['Quantity']).fillna(0).astype(int)


num_cols = [
    'Unit_Price_ZAR', 
    'Discount_Percent', 
    'Unit_Cost_ZAR', 
    'Revenue_ZAR', 
    'Profit_ZAR'
]

for col in num_cols:
    df[col] = pd.to_numeric(df[col])

print(df.dtypes)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df.to_csv('clean_retail_sales.csv', index=False, sep=';')
```
